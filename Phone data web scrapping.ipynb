```python
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
from urllib.request import urlopen 
from nltk import clean_html
import requests
```

```python
name = []
desc = []
price = []
rating =[]
```

```python
product_name_class = "_4rR01T"
product_description_class = "_1xgFaf"
product_price_class = "_30jeq3 _1_WHN1"
product_rating_class = "_3LWZlK"
```

```python
url = "https://www.flipkart.com/search?q=smartphone&sid=tyy%2C4io&as=on&as-show=on&otracker=AS_QueryStore_OrganicAutoSuggest_1_5_na_na_ps&otracker1=AS_QueryStore_OrganicAutoSuggest_1_5_na_na_ps&as-pos=1&as-type=RECENT&suggestionId=smartphone%7CMobiles&requestId=3e4ae282-5b03-4e89-9d45-c5ca869136cc&as-searchtext=smartphone&page=1"
r = requests.get(url)
soup = BeautifulSoup(r.content,'html.parser')
```

```python
product_name_1 = soup.find_all("div", {"class":product_name_class})
product_desc_1 = soup.find_all("ul",{"class":product_description_class})
product_price_1 = soup.find_all("div", {"class":product_price_class})
product_rating_1 = soup.find_all("div",{"class":product_rating_class})
```

```python
for j in range (1,41):
    url = "https://www.flipkart.com/search?q=smartphone&sid=tyy%2C4io&as=on&as-show=on&otracker=AS_QueryStore_OrganicAutoSuggest_1_5_na_na_ps&otracker1=AS_QueryStore_OrganicAutoSuggest_1_5_na_na_ps&as-pos=1&as-type=RECENT&suggestionId=smartphone%7CMobiles&requestId=3e4ae282-5b03-4e89-9d45-c5ca869136cc&as-searchtext=smartphone&page="+ str(j)
    r = requests.get(url)
    soup = BeautifulSoup(r.content,'html.parser')
    product_name_1 = soup.find_all("div", {"class":product_name_class})
    product_desc_1 = soup.find_all("ul",{"class":product_description_class})
    product_price_1 = soup.find_all("div", {"class":product_price_class})
    product_rating_1 = soup.find_all("div",{"class":product_rating_class})
    for i in range(0,23):
        temp1 = product_name_1[i].text.strip()
        temp2 = product_desc_1[i].text.strip()
        temp3 = product_price_1[i].text.strip()
        temp4 = product_rating_1[i].text.strip()
        name.append(temp1)
        desc.append(temp2)
        price.append(temp3)
        rating.append(temp4)
        
```

```python

df = pd.DataFrame(list(zip(name, price,rating)), columns =['name', 'price','rating'])
print (df)
```

```python
splits = []
for row in desc:
    split = row.split(" | ")
    splits.append(split)
```

```python
df2 = pd.DataFrame(splits, columns=["RAM", "ROM", "Expandable","Battery","Display","1","2"])
print(df2)
```

```python
final_df = pd.concat ([df, df2], axis=1)
print (final_df)
```

```python
final_df.to_excel("Phones classification.xlsx", index = False)
```

```python
url = "https://www.antutu.com/en/ranking/rank1.htm"
r = requests.get(url)
soup = BeautifulSoup(r.content,'html.parser')
```

```python
device = soup.find_all("li", {"class":device_name_class})
score = soup.find_all("li", {"class":device_score_class})
```
